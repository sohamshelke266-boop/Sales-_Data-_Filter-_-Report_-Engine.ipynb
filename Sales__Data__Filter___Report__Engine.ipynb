# Import NumPy and Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Create Month Labels

```python
months = np.arange(1, 13)

month_names = np.array([
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
])

print(months)
print(month_names)
```

# Generate Sales Data

```python
np.random.seed(10)

sales = np.random.randint(20000, 100001, 120).reshape(10, 12)

print(sales)
print("Sales data shape:", sales.shape)
```

# Slice Q1 and Q2

```python
q1 = sales[:, 0:3]
q2 = sales[:, 3:6]

print("Q1 Sales:")
print(q1)

print("Q2 Sales:")
print(q2)
```

# Slice Q3, Q4 and Individual Stores

```python
q3 = sales[:, 6:9]
q4 = sales[:, 9:12]

store1 = sales[0, :]
store5 = sales[4, :]

print("Q3 Sales:")
print(q3)

print("Q4 Sales:")
print(q4)

print("Store 1 Sales:")
print(store1)

print("Store 5 Sales:")
print(store5)
```

# Boolean Masking — Sales Above Target

```python
target = 60000

high_sales = sales > target

print("Sales above target:")
print(sales[high_sales])

print("Number of sales above target:", np.sum(high_sales))
```

# Below-Average Stores and Combined Conditions

```python
store_totals = np.sum(sales, axis=1)
average_sales = np.mean(store_totals)

below_average = store_totals < average_sales

print("Below-average stores:")
print(np.where(below_average)[0] + 1)

high_and_target = (sales > target) & (sales < 90000)
high_or_target = (sales > 90000) | (sales < 30000)
not_high_sales = ~(sales > target)

print("Combined AND condition:")
print(sales[high_and_target])

print("Combined OR condition:")
print(sales[high_or_target])

print("Sales not above target:")
print(sales[not_high_sales])
```

# Fancy Indexing — Select Stores

```python
selected_stores = sales[[1, 4, 7], :]

print("Selected stores: 2, 5 and 8")
print(selected_stores)
```

# Find the Best 3 Months

```python
monthly_totals = np.sum(sales, axis=0)

best_months = np.argsort(monthly_totals)[-3:][::-1]

print("Best 3 months:")
print(month_names[best_months])

print("Sales in best 3 months:")
print(monthly_totals[best_months])

best_month_sales = sales[np.ix_([1, 4, 7], best_months)]

print("Selected stores in best 3 months:")
print(best_month_sales)
```

# Growth Targets Using np.linspace

```python
growth_targets = np.linspace(50000, 100000, 12)

print("Monthly growth targets:")
print(growth_targets)
```

# Monthly and Store-Wise Reports

```python
monthly_totals = np.sum(sales, axis=0)
store_totals = np.sum(sales, axis=1)

print("Monthly Sales Report")

for i in range(12):
    print(month_names[i], monthly_totals[i], growth_targets[i])

print("Store-Wise Sales Report")

for i in range(10):
    print("Store", i + 1, "Total Sales:", store_totals[i])
```

# Visualize Sales Reports

```python
plt.figure(figsize=(10, 5))
plt.plot(month_names, monthly_totals, marker="o")
plt.plot(month_names, growth_targets, marker="x")
plt.title("Monthly Sales and Growth Targets")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend(["Actual Sales", "Growth Targets"])
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))
plt.bar(np.arange(1, 11), store_totals)
plt.title("Store-Wise Total Sales")
plt.xlabel("Store Number")
plt.ylabel("Total Sales")
plt.show()
```

# Product-Wise Sales Analysis

```python
products = np.array(["Laptop", "Mobile", "Tablet", "Headphones", "Keyboard"])

np.random.seed(20)

product_sales = np.random.randint(10000, 80001, 60).reshape(5, 12)

print("Product-Wise Monthly Sales")

for i in range(5):
    print(products[i], product_sales[i])

product_totals = np.sum(product_sales, axis=1)

print("Total Sales of Each Product")

for i in range(5):
    print(products[i], product_totals[i])

best_product = np.argmax(product_totals)

print("Best Selling Product:", products[best_product])
print("Highest Sales:", product_totals[best_product])
```
